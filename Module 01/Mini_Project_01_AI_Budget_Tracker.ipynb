### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE
monthly_income_str = input("Enter your monthly income: ")

try:
    monthly_income = float(monthly_income_str)
    if monthly_income < 0:
        print("Monthly income cannot be negative. Please enter a positive value.")
        # Exit or re-prompt, for this example we'll just stop.
    else:
        expenses = []
        for i in range(1, 4):
            expense_str = input(f"Enter expense {i} (or press Enter to skip): ")
            if expense_str:
                try:
                    expense = float(expense_str)
                    if expense < 0:
                        print("Expense cannot be negative. Skipping this expense.")
                    else:
                        expenses.append(expense)
                except ValueError:
                    print("Invalid expense amount. Skipping this expense.")

        total_expenses = sum(expenses)
        remaining_budget = monthly_income - total_expenses

        print(f"\n--- Budget Summary ---")
        print(f"Monthly Income: ${monthly_income:.2f}")
        for j, exp in enumerate(expenses):
            print(f"Expense {j+1}: ${exp:.2f}")
        print(f"Total Expenses: ${total_expenses:.2f}")
        print(f"Remaining Budget: ${remaining_budget:.2f}")

        if remaining_budget < 0:
            print("\n⚠️ Warning: Your budget is negative! You are spending more than you earn.")
        elif remaining_budget == 0:
            print("\nℹ️ Note: Your budget is perfectly balanced. Be mindful of unexpected costs.")
        else:
            print("\n✅ Good job! Your budget is positive.")

except ValueError:
    print("Invalid income entered. Please enter a numerical value for your monthly income.")
```

## Part 4: Enhancement Challenge

Pick ONE feature to add with AI's help:
- 🏷️ Expense categories (food, transport, entertainment)
- 📊 Percentage of income spent
- ✅ Input validation (no negative numbers)
- 📝 A detailed budget summary with advice


```python
# Document your enhancement choice and the prompt you used
enhancement_choice = "I chose to add: [a detailed budget summary with advice]"
enhancement_prompt = """
Enhance the budget tracker to provide a detailed budget summary. Calculate the percentage of income spent on expenses. Offer specific advice based on whether the remaining budget is positive, zero, or negative, including suggestions for saving, emergency funds, or cutting expenses.
"""

print(f"🚀 Enhancement: {enhancement_choice}")
```

```python
# ENHANCED BUDGET TRACKER
monthly_income_str = input("Enter your monthly income: ")

try:
    monthly_income = float(monthly_income_str)
    if monthly_income < 0:
        print("Monthly income cannot be negative. Please enter a positive value.")
    else:
        expenses = []
        for i in range(1, 4):
            expense_str = input(f"Enter expense {i} (or press Enter to skip): ")
            if expense_str:
                try:
                    expense = float(expense_str)
                    if expense < 0:
                        print("Expense cannot be negative. Skipping this expense.")
                    else:
                        expenses.append(expense)
                except ValueError:
                    print("Invalid expense amount. Skipping this expense.")

        total_expenses = sum(expenses)
        remaining_budget = monthly_income - total_expenses

        print(f"\n--- Detailed Budget Summary ---")
        print(f"Monthly Income: ${monthly_income:.2f}")
        for j, exp in enumerate(expenses):
            print(f"Expense {j+1}: ${exp:.2f}")
        print(f"Total Expenses: ${total_expenses:.2f}")
        print(f"Remaining Budget: ${remaining_budget:.2f}")

        if monthly_income > 0:
            percentage_spent = (total_expenses / monthly_income) * 100
            print(f"Percentage of income spent: {percentage_spent:.2f}%")

        print(f"\n--- Financial Advice ---")
        if remaining_budget < 0:
            print("⚠️ Warning: Your budget is negative! You are spending more than you earn.")
            print("Advice: It's crucial to review your expenses and identify areas where you can cut back. Prioritize essential spending and look for ways to increase your income.")
        elif remaining_budget == 0:
            print("ℹ️ Note: Your budget is perfectly balanced. Be mindful of unexpected costs.")
            print("Advice: Consider building an emergency fund to cover unforeseen expenses. Even small savings can make a big difference.")
        else:
            print("✅ Good job! Your budget is positive.")
            print("Advice: With a positive budget, you have opportunities to save, invest, or allocate funds towards your financial goals. Keep track of your spending to maintain this healthy financial position.")

except ValueError:
    print("Invalid income entered. Please enter a numerical value for your monthly income.")
```
